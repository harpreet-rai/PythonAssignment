```python
import seaborn as sns
import matplotlib.pyplot as plt

# You also need to make sure your 'data' variable is defined.
# If you are using the built-in tips dataset, run this:
data = sns.load_dataset('tips')
sns.regplot(x='total_bill', y='tip', data=data)
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', data=data,hue='time',style='smoker')
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', data=data,hue='smoker',col='time')
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', data=data,hue='time',style='smoker')#row wala
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', data=data,hue='time',style='day')
plt.show()
```

```python
sns.relplot(x='total_bill', y='tip', data=data,hue='time',col='day')
plt.show()
```

```python
sns.catplot(y='total_bill',x='day',data=data)
plt.show()
```

```python
sns.catplot(y='total_bill',x='day',data=data,kind='strip')
plt.show()
```
