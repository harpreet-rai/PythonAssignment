```python
import pandas as pd

# 1. File load karein (Bracket waala naam use karein jo explorer mein hai)
df = pd.read_csv("Used_Bikes (1).csv") 

# 2. Owner column ke liye dictionary
owner_dict = {
    'First Owner': 1,
    'Second Owner': 2,
    'Third Owner': 3,
    'Fourth Owner Or More': 4
}

# 3. Encoding apply karein
# .get() use karne se error nahi aayega agar koi value match na ho
df['owner'] = df['owner'].map(owner_dict)

# 4. Result check karein
df.head()
```
