🔄 Looping in Programming 🔄

Looping means repeating a set of instructions again and again until a certain condition is met. It’s like doing a dance step over and over until the music stops! 💃🎶

In short:

🔁 Loop = Repeat

**Why use loops?**

Because writing the same code many times is boring and messy! Loops make your program smart and efficient. 🧠✨

```python

# Print numbers from 1 to 5
for i in range(2,20,2):
    print("Number:", i)
```

```python
student={
    "name":'Harru',
    "class":'B.tech'
}
print(student)
```

```python
for key in student:
    print(key, ":", student[key])
```

```python
for x in range(1, 11):
    if x == 2:
        break
    print(x)
```

```python
1️⃣ break in Python
Definition:

break statement loop ko turant rok deta hai, chahe loop ka condition abhi true ho. 🔥
```

```python
for x in range(1, 6):
    if x == 4:
        break
    print(x)
```

2️⃣ continue in Python
Definition:

continue statement current iteration skip kar deta hai aur loop next iteration pe jump kar deta hai. ⚡

Syntax:
for item in collection:
    if condition:
        continue

```python
for x in range(1, 6):
    if x == 3:
        continue
    print(x)
```

🔁 While Loop in Python
Definition:

while loop ka use tab tak code repeat karne ke liye hota hai jab tak condition True rahe.

Think of it as: “Keep doing this until I say stop!” ⏳

```python
count = 1
while count <= 5:
    print("Count is:", count)
    count += 1
```
