The if statement in Python is a conditional control statement that is used to execute a block of code only when a given condition is true.
If the condition is false, the code inside the if block is skipped (or an alternative block like else / elif is executed).

In simple words:
👉 if is used to make decisions in a program.

Example:

```python
a =10
print(a)
```

```python
a = 10
b = 20

if a > b:
    print("a is greater")
else:
    print("b is greater")
```

```python

num = int(input("Enter a number: "))
if num % 2 == 0:
    print("Even number")
else:
    print("Odd number")
```
