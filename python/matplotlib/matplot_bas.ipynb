```python
import matplotlib.pyplot as plt

lst = [1, 2, 3, 4, 5]
label = ["co2", "N2", "O2", "NE", "Ar"]

plt.pie(lst, labels=label)
plt.show()
```

```python
import matplotlib.pyplot as plt   # matplotlib ka pyplot module import kar rahe hain

lst = [1, 2, 3, 4, 5]             # Pie chart ke slice sizes (values)
label = ["co2", "N2", "O2", "NE", "Ar"]  # Har slice ke labels

plt.pie(
    lst,
    labels=label,                 # Labels show karne ke liye
    autopct='%1.1f%%'             # Percentage display (1 decimal place)
)

plt.show()                        # Chart display karega
```

```python
lst=[1,2,3,4,5]
label=["co2","N2","O2","NE","Ar"]
my_explod=[0.2,0,0,0,0]
color=['red','green','blue','yellow','cyan']
plt.pie(lst,labels=label,shadow=True,startangle=90,explode=my_explod,colors=color,autopct='%1.1f%%')
plt.legend()
plt.show()
```
